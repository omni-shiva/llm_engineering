```python
sentences = [
    "Batman protects Gotham City.",
    "A superhero defends a city from criminals.",
    "A chef prepares pasta in an Italian restaurant."
]

embeddings = model.encode(sentences)

print("Type:", type(embeddings))
print("Number of NumPy axes:", embeddings.ndim)
print("Shape:", embeddings.shape)
print("Data type:", embeddings.dtype)
```

```python
print("Sentence 1:", sentences[0])
print("Embedding 1 shape:", embeddings[0].shape)

print("Sentence 2:", sentences[1])
print("Embedding 2 shape:", embeddings[1].shape)

print("Sentence 3:", sentences[2])
print("Embedding 3 shape:", embeddings[2].shape)
```

```python
from sentence_transformers.util import cos_sim

similarity_0_1 = cos_sim(embeddings[0], embeddings[1])
similarity_0_2 = cos_sim(embeddings[0], embeddings[2])

print("Batman vs superhero:", similarity_0_1)
print("Batman vs chef:", similarity_0_2)
```

```python
import numpy as np

A = np.array([2, 3])
B = np.array([4, 5])

dot_product = np.dot(A, B)

print("Dot product:", dot_product)
```

```python
magnitude_A = np.linalg.norm(A)
magnitude_B = np.linalg.norm(B)

print("Magnitude of A:", magnitude_A)
print("Magnitude of B:", magnitude_B)
```

```python
cosine_similarity = dot_product / (magnitude_A * magnitude_B)

print("Cosine similarity:", cosine_similarity)
```

```python
A = np.array([2.0, 3.0], dtype=np.float32)
B = np.array([4.0, 5.0], dtype=np.float32)

library_similarity = cos_sim(A, B)

print("Manual cosine similarity:", cosine_similarity)
print("Library cosine similarity:", library_similarity)
```

```python
batman = embeddings[0]
superhero = embeddings[1]
chef = embeddings[2]

batman_superhero_manual = np.dot(batman, superhero) / (
    np.linalg.norm(batman) * np.linalg.norm(superhero)
)

batman_chef_manual = np.dot(batman, chef) / (
    np.linalg.norm(batman) * np.linalg.norm(chef)
)

print("Batman vs superhero:", batman_superhero_manual)
print("Batman vs chef:", batman_chef_manual)
```

```python
movies = [
    "A masked superhero protects a dark city from dangerous criminals.",
    "A young lion must learn to become the king of his homeland.",
    "A chef travels across Italy learning traditional cooking techniques.",
    "A family of superheroes tries to balance ordinary life with saving the world.",
    "Two astronauts explore a mysterious planet far from Earth."
]
```

```python
query = "fun superhero movie about a family"
```

```python
import os
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
movies_embed = model.encode(movies)
```

```python
query_embed = model.encode(query)
```

```python
print("movie embed shape",movies_embed.shape)
print("query embed shape",query_embed.shape)

# (5,384) in shape means , 5embed * 384 vectors..
```

```python
scores=cos_sim(query_embed,movies_embed)
```

```python
print("Scores",scores)
print("Scores",scores.shape)
```

```python
scores_1d=scores[0]
ranking=scores_1d.argsort(descending=True)

print("Ranking indices:",ranking)

```
